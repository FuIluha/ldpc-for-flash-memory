# Симуляция канала с ошибками

+ $c = u \cdot G$, $c \in \{0, 1\}^n$ - сообщение
+ $r = с \oplus e $ - сообщение с ошибками 
+ $p$ - вероятность ошибки в канале на символ

```python
import numpy as np


def error_channel(c: np.array, p: float, n: int = 1) -> list:
    '''
    c --- сообщение
    p --- вероятность ошибки
    n --- кол-во повторов
    '''
    set_of_c = []
    for _ in range(n):
        error_vec = np.random.choice(2, len(c), p=[1. - p, p])
        c = (c + error_vec) % 2
        set_of_c.append(np.array(c))

    return set_of_c
```

```python
c = np.array(list(map(int, open("message.txt", "r").read().split(","))))
p = 0.01 # пока что вероятность константа
r = error_channel(c, p, n=100)
r
```

## Сохранение кодов с ошибками

```python
np.savetxt('error_messages.txt', np.array(r), fmt="%d", delimiter=',')
```
