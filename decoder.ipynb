## Объявления функций и подключение библиотек

```python
import numpy as np
import matplotlib.pyplot as plt

def dot_for_codes(a: np.array, b: np.array) -> np.array:
    return np.array(a @ b % 2)

def sum_for_codes(a: np.array) -> int:
    return np.sum(a) % 2
```

## Входные данные

+ $u \in \{0, 1\}^k$ - кодовое слово, в нашем случае, пока что нулевое кодовое слово
+ $c = u \cdot G$, $c \in \{0, 1\}^n$ - сообщение
+ $r = с \oplus e $ - сообщение с ошибками 
+ $p$ - вероятность ошибки в канале на символ
+ $H \in \{0, 1\}^{(n-k) \times k}$ - проверочная матрица, ее мы составляем сами
+ $G \in \{0, 1\}^{k \times n}$ - порождающая матрица пока, что мы ее получаем готовую
+ $S = H \cdot r^T$ - синдром, это вектор проверочных битов в графе Таннера

## Работа с данными

### Ввод проверочной матрицы 

```python
H = []
for vector in open("verification_matrix.txt", "r"):
    H.append(list(map(int, vector.split(","))))
H = np.array(H)
H.shape
```

### Ввод сообщений

```python
r = []
for vector in open("error_messages.txt", "r"):
    r.append(list(map(int, vector.split(","))))
r = np.array(r)
r.shape
```

Проверяем наличие ошибок в $r$ при помощи $H$ для этого выбираем те элементы кодового слова для каждой строки, с позициями ненулевых элементов $H$, после считаем сумму, это будет синдром

```python
S = []

for current_r in r:
    current_S = []

    for h in H.T:
        current_S.append(sum_for_codes(current_r[np.array(h, dtype=bool)]))

    S.append(current_S)

np.array(S)
```
