## Объявления функций и подключение библиотек

```python
import numpy as np
import matplotlib.pyplot as plt

def dot_for_codes(a: np.array, b: np.array) -> np.array:
    return np.array(a @ b % 2)

def sum_for_codes(a: np.array) -> int:
    return np.sum(a) % 2
```

## Входные данные

+ $u \in \{0, 1\}^k$ - кодовое слово, в нашем случае, пока что нулевое кодовое слово
+ $c = u \cdot G$, $c \in \{0, 1\}^n$ - сообщение
+ $G \in \{0, 1\}^{k \times n}$ - порождающая матрица пока, что мы ее получаем готовую

## Работа с данными

Порождающую матрицу, возьму такую

```python
G = []
for vector in open("matrix.txt", "r"):
    G.append(list(map(int, vector.split(","))))
G = np.array(G)
G.shape
```

```python
n = G.shape[1]
k = G.shape[0]
```

```python
u = np.zeros(k, dtype=int)
```

Кодируем сообщение

```python
c = dot_for_codes(u, G)
```

## Сохранение закодированного сообщения в файл

```python
file = open("message.txt", "w")
file.write(("".join(map(lambda x: str(x) + ",", c))[:-1]))
file.close()
```
