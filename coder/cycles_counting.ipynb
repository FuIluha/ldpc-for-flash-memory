# Поиск циклов в графе Таннера

```python
import numpy as np

H = np.loadtxt('../data/H_matrix/H3_1000x1500_fastPEG.txt', delimiter=',', dtype=np.int8)
# H = np.loadtxt('../data/matrix.txt', delimiter=',', dtype=np.int8)
```

```python
from numba import jit

def find_cycles_in_tanner_graph(H):
    rows, cols = H.shape
    cycles = {4: 0, 6: 0, 8: 0}
    
    indices = np.argwhere(H)
    graph = {i: set() for i in range(rows + cols)}
    
    for r, c in indices:
        graph[r].add(rows + c)
        graph[rows + c].add(r)
    
    @jit(nopython = True)
    def count_cycles(length):
        count = 0
        for start in range(rows + cols):
            paths = [[start]]
            while paths:
                path = paths.pop()
                last_node = path[-1]
                if len(path) == length:
                    if path[0] in graph[last_node]:
                        
                        count += 1
                    continue
                for neighbor in graph[last_node]:
                    if neighbor not in path or (len(path) == length - 1 and neighbor == path[0]):
                        paths.append(path + [neighbor])
        return count // (2 * length)  
    
    unique_cycles = set()
    for length in cycles.keys():
        cycles[length] = count_cycles(length)
    
    return cycles
```

```python
H = np.loadtxt('../data/H_matrix/H5_1000x1500.txt', delimiter=',', dtype=np.int8)
```

```python
с = find_cycles_in_tanner_graph(H)
с 
```
